# Artificial Intelligence, Machine Learning and Deep Learning

### Artificial Intelligence (AI)

**Definition:** Artificial Intelligence is the field of computer science focused on building systems that can perform tasks that normally require human intelligence such as reasoning, recognizing patterns, understanding language and making decisions.

### Machine Learning (ML)

**Definition:** Machine Learning is a branch of AI where computers learn patterns from historical data instead of being explicitly programmed with every rule.

Instead of telling the computer *how* to solve every problem, we provide examples and allow it to discover relationships in the data.

### Deep Learning (DL)

**Definition:** Deep Learning is a subset of Machine Learning that uses neural networks with many layers to learn complex patterns from large datasets.

Examples include ChatGPT, image recognition and speech recognition.

### Relationship

AI ⟶ Machine Learning ⟶ Deep Learning

Every Deep Learning model is a Machine Learning model, and every Machine Learning model belongs to Artificial Intelligence.


# What is Machine Learning?

### Definition

Machine Learning is the process of teaching a computer to learn from historical data so that it can make predictions or decisions about new, unseen data.

### Why do we need Machine Learning?

Many real-world problems contain thousands of variables. Writing rules for every situation is impossible.

Examples:
- Predicting stock prices
- Detecting fraud
- Recommending movies
- Predicting diseases

Machine Learning automatically discovers useful mathematical patterns hidden inside data.

### Traditional Programming

Rules + Data → Program → Answer

### Machine Learning

Historical Data + Correct Answers → Learning Algorithm → Model → Predictions

### Important Terms

**Algorithm:** A mathematical procedure used to learn patterns from data.

**Model:** The final learned representation produced by the algorithm. This is what makes predictions.


# Types of Machine Learning

## Supervised Learning

**Definition:** A learning method where the dataset contains both the inputs (features) and the correct outputs (labels).

Goal: Learn the relationship between features and labels.

Examples:
- House price prediction
- Stock price prediction
- Spam detection
- Disease prediction

Advantages:
- Easy to evaluate
- Usually accurate with quality labeled data

Limitations:
- Requires labeled datasets

---

## Unsupervised Learning

**Definition:** A learning method where only input data is provided. The algorithm must discover hidden patterns on its own.

Common tasks:
- Clustering
- Dimensionality Reduction

Examples:
- Customer segmentation
- Grouping similar stocks
- Recommendation systems

---

## Reinforcement Learning

**Definition:** A learning method where an agent interacts with an environment and learns through rewards and penalties.

Examples:
- Self-driving cars
- Trading agents
- Robotics
- Game playing


# Machine Learning Workflow

Every ML project generally follows the same pipeline.

1. Collect Data
2. Understand the Data
3. Clean the Data
4. Explore the Data (EDA)
5. Feature Engineering
6. Split the Dataset
7. Train the Model
8. Validate the Model
9. Test the Model
10. Deploy the Model
11. Monitor and Improve

### Definitions

**Data Cleaning:** Removing missing values, duplicates and incorrect data.

**Feature Engineering:** Creating better input variables that help the model learn.

**Deployment:** Making the trained model available for real users.

**Monitoring:** Continuously checking model performance after deployment.


# Important Machine Learning Terminology

### Dataset
A collection of data used for training or evaluating a model.

### Sample
One row in a dataset.

### Feature
An input variable used for prediction.

Examples:
- Daily Return
- Trading Volume
- Volatility

### Label (Target)
The value the model is trying to predict.

Examples:
- Tomorrow's Stock Price
- House Price
- Fraud / Not Fraud

### Prediction
The output generated by the trained model.


# Training, Validation and Testing

### Training Dataset

Used to teach the model by allowing it to learn patterns.

### Validation Dataset

Used while developing the model to compare different versions and tune parameters.

### Testing Dataset

Used only once after training is complete to estimate how well the model performs on unseen data.

### Why split the data?

If we evaluate the model using the same data it learned from, we cannot determine whether it truly understands the patterns or has simply memorized the data.


# Evaluation Metrics

Evaluation metrics measure the quality of a Machine Learning model.

## Regression Metrics

**MAE (Mean Absolute Error):**
Average absolute prediction error.

**MSE (Mean Squared Error):**
Squares each error before averaging, giving more importance to larger mistakes.

**RMSE (Root Mean Squared Error):**
Square root of MSE. Easier to interpret because it has the same units as the target.

**R² Score:**
Measures how much of the variation in the target variable is explained by the model.

## Classification Metrics

**Accuracy:** Percentage of correct predictions.

**Precision:** Out of all predicted positives, how many were actually positive?

**Recall:** Out of all actual positives, how many were correctly identified?

**F1 Score:** Harmonic mean of Precision and Recall.


# Overfitting and Underfitting

### Underfitting

Definition: The model is too simple to capture important relationships.

Characteristics:
- High training error
- High testing error

### Overfitting

Definition: The model memorizes the training data instead of learning general patterns.

Characteristics:
- Very low training error
- High testing error

### Generalization

The ability of a model to perform well on new, unseen data. This is the ultimate goal of Machine Learning.
